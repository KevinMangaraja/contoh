# hashing

1. Pengertian Hashing
Hashing adalah teknik untuk mengubah data (berupa teks atau angka) menjadi nilai numerik berukuran tetap yang disebut Hash Value atau Hash Code.
Teknik ini digunakan untuk menyimpan dan mencari data secara instan. Berbeda dengan pencarian biasa yang harus mengecek data satu per satu, hashing langsung tahu lokasi data tersebut disimpan.
2. Komponen Utama Hashing
- Hash Key: Data asli yang ingin disimpan atau dicari (contoh: Nama mahasiswa, nomor ID).
- Hash Function (Fungsi Hash): Rumus matematika yang mengubah Hash Key menjadi indeks angka.
- Hash Table: Array atau tempat penyimpanan data akhir, di mana posisi datanya ditentukan oleh indeks hasil Hash Function.
3. Masalah Utama: Collision (Tabrakan Data)
Collision terjadi ketika ada dua data berbeda, namun saat dimasukkan ke dalam Hash Function, menghasilkan nilai indeks yang sama.

Cara Mengatasi Collision (Collision Resolution):
- Chaining (Open Addressing alternatif): Setiap indeks di Hash Table tidak langsung berisi data, melainkan berisi sebuah Linked List. Jika terjadi tabrakan, data baru cukup ditambahkan di belakang data lama pada indeks yang sama.
- Open Addressing (Linear Probing): Jika indeks yang dituju sudah terisi, program akan mencari indeks kosong berikutnya secara berurutan di bawahnya (maju +1).

```python
class HashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def _hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self._hash_function(key)
        start_index = index
        while self.table[index] is not None:
            if self.table[index][0] == key:
                self.table[index] = (key, value)
                return
            index = (index + 1) % self.size
            if index == start_index:
                return

        self.table[index] = (key, value)

    def search(self, key):
        index = self._hash_function(key)
        start_index = index
        while self.table[index] is not None:
            if self.table[index][0] == key:
                return self.table[index][1]
            index = (index + 1) % self.size
            if index == start_index:
                break
        return None

# --- UJI COBA KODE ---
ht = HashTable(5)
ht.insert("Budi", "Nilai A")
ht.insert("Andi", "Nilai B")

print(ht.search("Budi"))
print(ht.search("Andi"))
```
