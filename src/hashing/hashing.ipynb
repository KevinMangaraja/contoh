```python
# A. Close Addressing (Chaining) + CRUD
# CLOSE ADDRESSING (CHAINING)

class HashTable:
    def __init__(self, size):
        self.size = size
        self.table = [[] for _ in range(size)]

    # fungsi hash
    def hash_function(self, key):
        return key % self.size

    # CREATE / INSERT
    def insert(self, key):
        index = self.hash_function(key)

        if key not in self.table[index]:
            self.table[index].append(key)
            print(f"{key} berhasil ditambahkan")
        else:
            print(f"{key} sudah ada")

    # READ / SEARCH
    def search(self, key):
        index = self.hash_function(key)

        if key in self.table[index]:
            print(f"{key} ditemukan pada index {index}")
        else:
            print(f"{key} tidak ditemukan")

    # UPDATE
    def update(self, old_key, new_key):
        index = self.hash_function(old_key)

        if old_key in self.table[index]:
            self.table[index].remove(old_key)

            new_index = self.hash_function(new_key)
            self.table[new_index].append(new_key)

            print(f"{old_key} diganti menjadi {new_key}")
        else:
            print(f"{old_key} tidak ditemukan")

    # DELETE
    def delete(self, key):
        index = self.hash_function(key)

        if key in self.table[index]:
            self.table[index].remove(key)
            print(f"{key} berhasil dihapus")
        else:
            print(f"{key} tidak ditemukan")

    # DISPLAY
    def display(self):
        print("\nIsi Hash Table:")
        for i, value in enumerate(self.table):
            print(i, ":", value)


# PROGRAM UTAMA
ht = HashTable(5)

data = [10, 15, 20, 7, 12]

for item in data:
    ht.insert(item)

ht.display()

ht.search(15)

ht.update(15, 99)

ht.display()

ht.delete(20)

ht.display()
```

```python
# B. Open Addressing – Linear Probing + CRUD
# OPEN ADDRESSING - LINEAR PROBING

class LinearProbing:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return key % self.size

    # CREATE
    def insert(self, key):

        # LOAD FACTOR CHECK
        if None not in self.table:
            print("Hash Table Penuh")
            return

        index = self.hash_function(key)

        while self.table[index] is not None:
            index = (index + 1) % self.size

        self.table[index] = key
        print(f"{key} masuk index {index}")

    # READ
    def search(self, key):
        index = self.hash_function(key)
        start = index

        while self.table[index] is not None:

            if self.table[index] == key:
                print(f"{key} ditemukan di index {index}")
                return

            index = (index + 1) % self.size

            if index == start:
                break

        print(f"{key} tidak ditemukan")

    # UPDATE
    def update(self, old, new):
        self.delete(old)
        self.insert(new)

    # DELETE
    def delete(self, key):
        index = self.hash_function(key)
        start = index

        while self.table[index] is not None:

            if self.table[index] == key:
                self.table[index] = None
                print(f"{key} dihapus")
                return

            index = (index + 1) % self.size

            if index == start:
                break

        print(f"{key} tidak ditemukan")

    def display(self):
        print(self.table)


# PROGRAM
lp = LinearProbing(5)

data = [10, 15, 20, 25, 30]

for x in data:
    lp.insert(x)

lp.display()

# contoh penuh
lp.insert(35)

lp.search(20)

lp.update(20, 99)

lp.display()

lp.delete(15)

lp.display()
```

```python
# C. Open Addressing – Quadratic Probing
# QUADRATIC PROBING

class QuadraticProbing:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return key % self.size

    def insert(self, key):

        if None not in self.table:
            print("Hash Table Penuh")
            return

        index = self.hash_function(key)

        i = 0

        while self.table[(index + i*i) % self.size] is not None:
            i += 1

        new_index = (index + i*i) % self.size

        self.table[new_index] = key

        print(f"{key} masuk index {new_index}")

    def display(self):
        print(self.table)


# PROGRAM
qp = QuadraticProbing(7)

data = [10, 17, 24, 31]

for x in data:
    qp.insert(x)

qp.display()
```

```python
# D. Mid Square Method (Fungsi Hash)
# MID SQUARE METHOD

def mid_square_hash(key, size):

    square = key * key

    square_str = str(square)

    mid = len(square_str) // 2

    # ambil 2 digit tengah
    middle_digits = square_str[mid-1:mid+1]

    hash_value = int(middle_digits) % size

    return hash_value


# TEST
size = 10

keys = [12, 44, 56, 78]

for key in keys:
    print(f"Key {key} -> Hash {mid_square_hash(key, size)}")
```
