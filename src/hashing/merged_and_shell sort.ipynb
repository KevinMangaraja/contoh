# Merge Sort and shell sort

# Pengertian merge sort

Merge Sort adalah algoritma pengurutan (sorting) yang menggunakan metode Divide and Conquer.
Algoritma ini bekerja dengan cara:

- Membagi data menjadi dua bagian terus menerus.
- Mengurutkan masing-masing bagian.
- Menggabungkan kembali bagian-bagian tersebut menjadi data yang sudah urut.

Merge Sort sangat cocok digunakan untuk data yang besar karena prosesnya stabil dan efisien.

Cara Kerja Merge Sort :

Contoh data:

    [38, 27, 43, 3, 9, 82, 10]
    Langkah 1 — Membagi Data
    [38, 27, 43, 3]      [9, 82, 10]

    Dibagi lagi:

    [38, 27] [43, 3] [9, 82] [10]

    Dibagi sampai menjadi satu elemen:

    [38] [27] [43] [3] [9] [82] [10]
    Langkah 2 — Menggabungkan dan Mengurutkan

    Gabungkan sambil dibandingkan:

    [27, 38]
    [3, 43]
    [9, 82]
    [10]

    Gabung lagi:

    [3, 27, 38, 43]
    [9, 10, 82]

    Hasil akhir:

    [3, 9, 10, 27, 38, 43, 82]

Kelebihan Merge Sort:
- Cepat untuk data besar
- Stabil (data yang sama tidak berubah posisi)
- Kompleksitas waktu konsisten

Kekurangan Merge Sort
- Membutuhkan memori tambahan
- Lebih rumit dibanding Bubble Sort

contoh code merged sort :


```python
def merge_sort(arr):
    if len(arr) > 1:

        # Membagi array menjadi dua
        mid = len(arr) // 2
        left = arr[:mid]
        right = arr[mid:]

        # Rekursif untuk membagi lagi
        merge_sort(left)
        merge_sort(right)

        i = j = k = 0

        # Menggabungkan data yang sudah diurutkan
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1
            else:
                arr[k] = right[j]
                j += 1
            k += 1

        # Sisa elemen kiri
        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1

        # Sisa elemen kanan
        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1


# Contoh penggunaan
data = [38, 27, 43, 3, 9, 82, 10]

print("Data sebelum sorting:")
print(data)

merge_sort(data)

print("Data setelah sorting:")
print(data)
```

# Shell Sort

Pengertian

Shell Sort adalah pengembangan dari Insertion Sort.
Algoritma ini bekerja dengan membandingkan elemen yang memiliki jarak tertentu (gap), lalu gap diperkecil secara bertahap sampai menjadi 1.

Tujuannya agar proses sorting lebih cepat dibanding Insertion Sort biasa.

Cara Kerja Shell Sort

    Contoh data:

    [23, 12, 1, 8, 34, 54, 2, 3]
    Langkah 1 — Menentukan Gap

    Misal:

    gap = jumlah data / 2
    gap = 8 / 2 = 4

    Bandingkan elemen yang berjarak 4:

    23 ↔ 34
    12 ↔ 54
    1 ↔ 2
    8 ↔ 3

    Hasil:

    [23, 12, 1, 3, 34, 54, 2, 8]
    Langkah 2 — Gap Diperkecil

    Gap berikutnya:

    gap = 2

    Bandingkan lagi.

    Langkah 3 — Gap Menjadi 1

    Saat gap = 1, proses seperti Insertion Sort biasa.

    Hasil akhir:

    [1, 2, 3, 8, 12, 23, 34, 54]

Kelebihan Shell Sort
- Lebih cepat dari Bubble Sort dan Insertion Sort
- Mudah dipahami
- Tidak membutuhkan memori tambahan besar

Kekurangan Shell Sort
- Tidak stabil
- Performa tergantung pemilihan gap

```python
def shell_sort(arr):
    n = len(arr)
    gap = n // 2

    while gap > 0:

        for i in range(gap, n):

            temp = arr[i]
            j = i

            while j >= gap and arr[j - gap] > temp:
                arr[j] = arr[j - gap]
                j -= gap

            arr[j] = temp

        gap //= 2


data = [23, 12, 1, 8, 34, 54, 2, 3]

print("Data sebelum sorting:")
print(data)

shell_sort(data)

print("Data setelah sorting:")
print(data)
```

Perbedaan Algoritma Sorting

| Algoritma          | Cara Kerja                              | Kompleksitas | Kelebihan                  | Kekurangan                      |
| ------------------ | --------------------------------------- | ------------ | -------------------------- | ------------------------------- |
| **Bubble Sort**    | Menukar elemen bersebelahan sampai urut | O(n²)        | Mudah dipahami             | Sangat lambat                   |
| **Selection Sort** | Memilih nilai terkecil lalu ditukar     | O(n²)        | Pertukaran sedikit         | Tetap lambat                    |
| **Insertion Sort** | Menyisipkan data ke posisi yang benar   | O(n²)        | Cepat untuk data kecil     | Kurang efektif untuk data besar |
| **Quick Sort**     | Membagi data berdasarkan pivot          | O(n log n)   | Sangat cepat               | Worst case O(n²)                |
| **Merge Sort**     | Membagi lalu menggabungkan data terurut | O(n log n)   | Stabil dan cepat           | Butuh memori tambahan           |
| **Shell Sort**     | Insertion Sort dengan jarak (gap)       | O(n²)        | Lebih cepat dari insertion | Tidak stabil                    |
