# Tree

Pengertian

Tree adalah struktur data berbentuk hirarki yang terdiri dari node (simpul) dan edge (cabang/penghubung).

Struktur tree mirip seperti pohon terbalik:
- Akar berada di atas
- Cabang menyebar ke bawah

Struktur Dasar Tree

         A
       /   \
      B     C
     / \   / \
    D   E F   G

| Istilah | Penjelasan             |
| ------- | ---------------------- |
| Root    | Node paling atas       |
| Parent  | Node induk             |
| Child   | Node anak              |
| Leaf    | Node tanpa anak        |
| Edge    | Penghubung antar node  |
| Subtree | Bagian kecil dari tree |
| Level   | Tingkatan node         |


1. Binary Tree

Setiap node maksimal memiliki 2 anak.

      A
     / \
    B   C

2. Binary Search Tree (BST)

Binary tree dengan aturan:
- Kiri lebih kecil
- Kanan lebih besar

      8
     / \
    3   10

3. Full Binary Tree

Setiap node memiliki:

- 0 anak, atau
- 2 anak

4. Complete Binary Tree

Semua level penuh kecuali level terakhir.

Traversal Tree

    Traversal = cara mengunjungi node.

    1. Preorder

    Root → Kiri → Kanan

    A B D E C F G
    2. Inorder

    Kiri → Root → Kanan

    D B E A F C G
    3. Postorder

    Kiri → Kanan → Root

    D E B F G C A
