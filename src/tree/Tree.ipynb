```python
# menampilkan tree
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def print_tree(root, level=0):
    if root:
        print_tree(root.right, level + 1)
        print("    " * level + str(root.data))
        print_tree(root.left, level + 1)


# Membuat Tree
root = Node("A")
root.left = Node("B")
root.right = Node("C")
root.left.left = Node("D")
root.left.right = Node("E")
root.right.left = Node("F")
root.right.right = Node("G")

print("=== TREE ===")
print_tree(root)
```

```python
# 2 Menampilkan program trace parse tree, membaca token, current node, menampilkan Parse Tree
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def print_tree(root, level=0):
    if root:
        print_tree(root.right, level + 1)
        print("    " * level + str(root.data))
        print_tree(root.left, level + 1)


# Input ekspresi
expr = "A + B * C"

print("=== TRACE PARSE TREE ===\n")

# Membaca token
tokens = expr.split()

for i, token in enumerate(tokens, start=1):
    print(f"Langkah {i}")
    print("Membaca Token :", token)
    print("Current Node  :", token)
    print("-" * 25)

# Membentuk Parse Tree
# A + (B * C)

root = Node("+")
root.left = Node("A")

root.right = Node("*")
root.right.left = Node("B")
root.right.right = Node("C")

print("\n=== HASIL PARSE TREE ===")
print_tree(root)
```
