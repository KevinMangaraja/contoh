A. Soal Pemrograman Stack

    1. Buat program stack dengan operasi push, pop, peek, isEmpty, size.
    2. Tentukan isi stack setelah operasi: push(10), push(20), push(30), pop(), push(40).
    3. Buat program membalik kata menggunakan stack.
    4. Cek keseimbangan tanda kurung menggunakan stack.
    5. Tampilkan langkah perubahan isi stack.
    6. Stack dengan batas maksimum (overflow).
    7. Jelaskan perbedaan peek() dan pop().
    8. Hitung jumlah elemen genap dalam stack.

jawab : 

```python
#1.
class Stack:
    def __init__(self, max_size=None):
        self.stack = []
        self.max_size = max_size

    def isEmpty(self):
        return len(self.stack) == 0

    def size(self):
        return len(self.stack)

    def push(self, item):
        if self.max_size is not None and len(self.stack) >= self.max_size:
            print("Stack Overflow!")
        else:
            self.stack.append(item)
            print(f"Push {item} -> {self.stack}")

    def pop(self):
        if self.isEmpty():
            print("Stack Underflow!")
            return None
        item = self.stack.pop()
        print(f"Pop {item} -> {self.stack}")
        return item

    def peek(self):
        if self.isEmpty():
            return None
        return self.stack[-1]
#2.
##penggunaan
# push(10), push(20), push(30), pop(), push(40)

#3. 
def balik_kata(kata):
    s = []
    for huruf in kata:
        s.append(huruf)

    hasil = ""
    while s:
        hasil += s.pop()

    return hasil

# Contoh
print(balik_kata("python"))  # nohtyp

#4. 
def cek_kurung(expr):
    stack = []
    pasangan = {')': '(', '}': '{', ']': '['}

    for char in expr:
        if char in "({[":
            stack.append(char)
        elif char in ")}]":
            if not stack or stack.pop() != pasangan[char]:
                return False

    return len(stack) == 0

# Contoh
print(cek_kurung("({[]})"))  # True
print(cek_kurung("({[})"))   # False

# 5. 
# push(5)  → [5]
# push(7)  → [5, 7]
# push(9)  → [5, 7, 9]
# pop()    → [5, 7]
# peek()   → 7 (tidak menghapus)

# 6. 
s = Stack(max_size=3)

s.push(1)
s.push(2)
s.push(3)
s.push(4)  # Overflow

# 7.
# Perbedaan peek() dan pop()
# | Fitur          | peek()             | pop()              |
# | -------------- | ------------------ | ------------------ |
# | Mengambil data | Ya                 | Ya                 |
# | Menghapus data | Tidak              | Ya                 |
# | Posisi         | Elemen paling atas | Elemen paling atas |

#8. 
def hitung_genap(stack):
    count = 0
    for item in stack:
        if item % 2 == 0:
            count += 1
    return count

# Contoh
s = [10, 15, 20, 25, 30]
print("Jumlah genap:", hitung_genap(s))  # 3
```

```python
#1. 
class Queue:
    def __init__(self, max_size=None):
        self.queue = []
        self.max_size = max_size

    def isEmpty(self):
        return len(self.queue) == 0

    def size(self):
        return len(self.queue)

    def enqueue(self, item):
        if self.max_size is not None and len(self.queue) >= self.max_size:
            print("Queue Overflow!")
        else:
            self.queue.append(item)
            print(f"Enqueue {item} -> {self.queue}")

    def dequeue(self):
        if self.isEmpty():
            print("Queue Underflow!")
            return None
        item = self.queue.pop(0)
        print(f"Dequeue {item} -> {self.queue}")
        return item

    def front(self):
        if self.isEmpty():
            return None
        return self.queue[0]
# 2.
# operasi:
# enqueue(10), enqueue(20), enqueue(30), dequeue(), enqueue(40)
# langkah :
# []
# enqueue(10) → [10]
# enqueue(20) → [10, 20]
# enqueue(30) → [10, 20, 30]
# dequeue()   → [20, 30]
# enqueue(40) → [20, 30, 40]

#3. 
q = Queue()

q.enqueue("Pasien A")
q.enqueue("Pasien B")
q.enqueue("Pasien C")

q.dequeue()  # Pasien A dilayani
q.enqueue("Pasien D")

print("Antrian sekarang:", q.queue)

# output:
# ['Pasien B', 'Pasien C', 'Pasien D']

# 4..
# FIFO dan Perbedaan dengan Stack
# Queue (FIFO = First In First Out)
# → Yang masuk duluan, keluar duluan
# Contoh: antrian pasien
# Stack (LIFO = Last In First Out)
# → Yang masuk terakhir, keluar duluan
# Contoh: tumpukan buku

# | Struktur | Cara Kerja |
# | -------- | ---------- |
# | Queue    | FIFO       |
# | Stack    | LIFO       |

# 5. 
def tampil_queue(q):
    print("Isi queue:", q.queue)

# Contoh
q = Queue()
q.enqueue(1)
q.enqueue(2)
tampil_queue(q)

# 6.
# operasi untuk fungsi 5
# enqueue(5), enqueue(10), dequeue(), enqueue(15), dequeue(), enqueue(20)

# langkah:
# []
# enqueue(5)  → [5]
# enqueue(10) → [5, 10]
# dequeue()   → [10]
# enqueue(15) → [10, 15]
# dequeue()   → [15]
# enqueue(20) → [15, 20]

7.
q = Queue(max_size=3)

q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
q.enqueue(4)  # Overflow

# output:
# Enqueue 1 -> [1]
# Enqueue 2 -> [1, 2]
# Enqueue 3 -> [1, 2, 3]
# Queue Overflow!

# 8.
# enqueque:
# IF queue penuh THEN
#     tampilkan "Overflow"
# ELSE
#     tambahkan elemen ke belakang queue
# END IF

# dequeque:
# IF queue kosong THEN
#     tampilkan "Underflow"
# ELSE
#     ambil elemen paling depan
#     hapus elemen tersebut dari queue
# END IF
```

C. Soal Pemrograman Deque

    1. Buat program deque dengan addFront, addRear, removeFront, removeRear.
    2. Jelaskan konsep deque.
    3. Tentukan isi deque setelah operasi tertentu.
    4. Simulasi antrian prioritas.
    5. Perbedaan queue dan deque.
    6. Cek palindrome menggunakan deque.
    7. Tentukan isi akhir deque.

```python
# 1.
class Deque:
    def __init__(self):
        self.deque = []

    def isEmpty(self):
        return len(self.deque) == 0

    def addFront(self, item):
        self.deque.insert(0, item)
        print(f"addFront({item}) -> {self.deque}")

    def addRear(self, item):
        self.deque.append(item)
        print(f"addRear({item}) -> {self.deque}")

    def removeFront(self):
        if self.isEmpty():
            print("Deque kosong!")
            return None
        item = self.deque.pop(0)
        print(f"removeFront() -> {item}, sisa {self.deque}")
        return item

    def removeRear(self):
        if self.isEmpty():
            print("Deque kosong!")
            return None
        item = self.deque.pop()
        print(f"removeRear() -> {item}, sisa {self.deque}")
        return item

    def size(self):
        return len(self.deque)

# 2.
# Deque (Double-Ended Queue) adalah struktur data yang memungkinkan:

# Penambahan elemen dari depan (front) dan belakang (rear)
# Penghapusan elemen dari dua sisi

# 3.
# operasi:
# addRear(10), addRear(20), addFront(5), removeRear(), addFront(1)

# langkah:
# []
# addRear(10)  → [10]
# addRear(20)  → [10, 20]
# addFront(5)  → [5, 10, 20]
# removeRear() → [5, 10]
# addFront(1)  → [1, 5, 10]

# output:
# [1, 5, 10]

# 4.
# Simulasi Antrian Prioritas

# Misal:

# VIP → masuk depan
# Biasa → masuk belakang

# d = Deque()

# d.addRear("Pasien Biasa 1")
# d.addRear("Pasien Biasa 2")
# d.addFront("Pasien VIP 1")  # prioritas
# d.addFront("Pasien VIP 2")

# d.removeFront()  # VIP dilayani dulu

# print("Antrian sekarang:", d.deque)

# output:
# ['Pasien VIP 1', 'Pasien Biasa 1', 'Pasien Biasa 2']

# 5.
# Perbedaan Queue dan Deque:

# | Fitur         | Queue         | Deque            |
# | ------------- | ------------- | ---------------- |
# | Akses         | Satu arah     | Dua arah         |
# | Tambah data   | Belakang saja | Depan & belakang |
# | Hapus data    | Depan saja    | Depan & belakang |
# | Fleksibilitas | Terbatas      | Lebih fleksibel  |

# 6.
# Cek Palindrome Menggunakan Deque:

def palindrome(kata):
    d = []

    for huruf in kata:
        d.append(huruf)

    while len(d) > 1:
        if d.pop(0) != d.pop():
            return False

    return True

# Contoh
print(palindrome("katak"))  # True
print(palindrome("python")) # False

# 7.
# operasi:
# addFront(3), addRear(6), addFront(1), removeFront(), addRear(9), removeRear()

# langkah:

# []
# addFront(3)  → [3]
# addRear(6)   → [3, 6]
# addFront(1)  → [1, 3, 6]
# removeFront()→ [3, 6]
# addRear(9)   → [3, 6, 9]
# removeRear() → [3, 6]

# output:
# [3, 6]
```

2.Program Bubble Sort Ascending

def bubble_sort_asc(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

    print(bubble_sort_asc([5,3,8,4]))

def bubble_sort_swap(arr):
    n = len(arr)
    swap = 0

    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]

5.Konsep Bubble Sort

Bubble Sort adalah metode pengurutan dengan cara:

Membandingkan dua elemen bersebelahan
Menukar jika urutannya salah
Elemen terbesar akan “menggelembung” ke akhir (seperti bubble)

Ciri:

Sederhana
Kurang efisien untuk data besar
Kompleksitas: O(n²)

6. Bubble Sort dengan Optimasi

Optimasi: berhenti jika tidak ada pertukaran

def bubble_sort_optimasi(arr):
    n = len(arr)

    for i in range(n):
        swapped = False

        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True

        if not swapped:
            break

    return arr

    print(bubble_sort_optimasi([1,2,3,4]))

2. Program Insertion Sort Ascending

        def insertion_sort_asc(arr):
            for i in range(1, len(arr)):
                key = arr[i]
                j = i - 1

                while j >= 0 and arr[j] > key:
                    arr[j + 1] = arr[j]
                    j -= 1

                arr[j + 1] = key

            return arr

        print(insertion_sort_asc([5,3,8,4]))

3. Insertion Sort Descending

        def insertion_sort_desc(arr):
            for i in range(1, len(arr)):
                key = arr[i]
                j = i - 1

                while j >= 0 and arr[j] < key:
                    arr[j + 1] = arr[j]
                    j -= 1

                arr[j + 1] = key

            return arr

        print(insertion_sort_desc([5,3,8,4]))

4. Analogi Kartu

    Insertion Sort seperti menyusun kartu di tangan:

    Ambil satu kartu
    Bandingkan dengan kartu sebelumnya
    Sisipkan di posisi yang benar

    Contoh:

    [5,3] → ambil 3 → sisipkan sebelum 5 → [3,5]

5. Proses Pergeseran Elemen

Saat elemen tidak pada posisi yang benar:

Elemen yang lebih besar digeser ke kanan
Ruang kosong digunakan untuk menyisipkan elemen

    Contoh:

    [3,5,8,4]

    → geser 8 → [3,5,4,8]

    → geser 5 → [3,4,5,8]

6. Perbedaan dengan Selection Sort

| Aspek      | Insertion Sort                     | Selection Sort |
| ---------- | ---------------------------------- | -------------- |
| Cara kerja | Sisipkan elemen                    | Pilih minimum  |
| Pergerakan | Banyak geser                       | Sedikit swap   |
| Efisiensi  | Lebih cepat untuk data hampir urut | Tetap O(n²)    |
| Stabil     | Ya                                 | Tidak          |

8. Kapan Digunakan?

    Insertion Sort cocok digunakan saat:

    Data kecil
    Data hampir terurut
    Butuh algoritma sederhana & stabil

    ❌ Tidak cocok untuk:

    Data besar (karena O(n²))
