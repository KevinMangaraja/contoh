```python
import random
import time


# BUBBLE SORT
def bubble_sort(arr):
    a = arr.copy()

    for i in range(len(a)):
        for j in range(len(a)-1-i):

            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]

    return a


# SELECTION SORT
def selection_sort(arr):
    a = arr.copy()

    for i in range(len(a)):

        min_idx = i

        for j in range(i+1, len(a)):

            if a[j] < a[min_idx]:
                min_idx = j

        a[i], a[min_idx] = a[min_idx], a[i]

    return a


# INSERTION SORT
def insertion_sort(arr):
    a = arr.copy()

    for i in range(1, len(a)):

        key = a[i]
        j = i - 1

        while j >= 0 and a[j] > key:
            a[j+1] = a[j]
            j -= 1

        a[j+1] = key

    return a


# MERGE SORT
def merge_sort(arr):

    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    result = []

    i = j = 0

    while i < len(left) and j < len(right):

        if left[i] < right[j]:
            result.append(left[i])
            i += 1

        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


# QUICK SORT
def quick_sort(arr):

    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr)//2]

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + right


# PROGRAM UTAMA
sizes = [100, 200, 300, 400, 500]

for size in sizes:

    data = random.sample(range(1, 1000), size)

    print(f"\nJumlah Data = {size}")

    start = time.time()
    bubble_sort(data)
    end = time.time()
    print("Bubble Sort   :", end - start)

    start = time.time()
    selection_sort(data)
    end = time.time()
    print("Selection Sort:", end - start)

    start = time.time()
    insertion_sort(data)
    end = time.time()
    print("Insertion Sort:", end - start)

    start = time.time()
    merge_sort(data)
    end = time.time()
    print("Merge Sort    :", end - start)

    start = time.time()
    quick_sort(data)
    end = time.time()
    print("Quick Sort    :", end - start)
```
