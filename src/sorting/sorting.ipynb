1. Bubble Sort (Pengurutan Gelembung)Algoritma yang mengurutkan data dengan cara membandingkan dua elemen yang bersebelahan secara berulang, lalu menukarnya jika urutannya salah. Elemen terbesar akan "mengapung" ke posisi akhir seperti gelembung.
- Kelebihan: Sangat sederhana untuk dipahami.
- Kekurangan: Tidak efisien untuk data dalam jumlah besar.

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

print(bubble_sort([64, 34, 25, 12, 22, 11, 90]))
```

2. Selection Sort (Pengurutan Pilihan)
Algoritma ini bekerja dengan cara mencari elemen terkecil dari bagian data yang belum diurutkan, lalu menukarnya dengan elemen pertama di bagian tersebut. Proses ini diulangi untuk posisi berikutnya.
- Kelebihan: Jumlah pertukaran data (swap) minimal.
- Kekurangan: Kecepatannya tetap lambat meskipun data sudah hampir terurut.

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

print(selection_sort([64, 25, 12, 22, 11]))
```

3. Insertion Sort (Pengurutan Sisipan)
Algoritma ini bekerja seperti cara manusia mengurutkan kartu di tangan. Data diambil satu per satu dari bagian yang belum terurut, lalu disisipkan ke posisi yang tepat pada bagian yang sudah terurut.
- Kelebihan: Sangat cepat jika data inputnya sudah hampir terurut.
- Kekurangan: Kurang efisien jika urutan data awal acak total atau terbalik.

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

print(insertion_sort([12, 11, 13, 5, 6]))
```

4. Quick Sort (Pengurutan Cepat)
Algoritma modern yang menggunakan pendekatan Divide and Conquer (bagi dan kuasai). Algoritma ini memilih satu elemen sebagai Pivot, lalu membagi data menjadi dua bagian: elemen yang lebih kecil dari pivot di sebelah kiri, dan yang lebih besar di sebelah kanan. Proses dilakukan berulang secara rekursif.
- Kelebihan: Sangat cepat dan paling sering digunakan dalam praktik nyata.
- Kekurangan: Implementasinya lebih kompleks karena menggunakan fungsi rekursif.

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

print(quick_sort([3, 6, 8, 10, 1, 2, 1]))
```
