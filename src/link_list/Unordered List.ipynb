```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class UnorderedList:
    def __init__(self):
        self.head = None

    # tambah di akhir
    def append(self, data):
        newNode = Node(data)

        if self.head is None:
            self.head = newNode
            return

        temp = self.head
        while temp.next:
            temp = temp.next

        temp.next = newNode

    # insert sesuai aturan
    def insert_after_search(self, item):
        temp = self.head

        while temp:
            # jika ditemukan
            if temp.data == item:
                newNode = Node(item)

                newNode.next = temp.next
                temp.next = newNode
                return

            temp = temp.next

        # jika tidak ditemukan → tambah di akhir
        self.append(item)

    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end=" , ")
            temp = temp.next

        print("None")


# Program utama
ulist = UnorderedList()

ulist.append(7)
ulist.append(18)
ulist.append(6)

print("Sebelum:")
ulist.display()

ulist.insert_after_search(18)

print("Sesudah:")
ulist.display()
```
