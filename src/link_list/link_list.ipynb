# Linked List

## 1.Pengertian Linked List

Linked List adalah struktur data linear yang terdiri dari kumpulan node yang saling terhubung. Setiap node memiliki:

Data → menyimpan nilai.
Pointer (link) → menyimpan alamat node berikutnya.

Berbeda dengan array yang menyimpan data secara berurutan di memori, linked list menyimpan data secara dinamis dan saling terhubung menggunakan pointer.

Ilustrasi

        Head
        |
        v
        +------+------+    +------+------+    +------+
        |  10  |  o--|--->|  20  |  o--|--->|  30  | None
        +------+------+    +------+------+    +------+


## 2.Komponen Linked List

a. Node

Node adalah elemen dasar dalam linked list yang berisi:

Data: 
Pointer ke node berikutnya

Contoh:

    class Node:
        def __init__(self, data):
            self.data = data
            self.next = None

b. Head

Head adalah node pertama dalam linked list.

Contoh:

    Head → 10 → 20 → 30 → None

## 3. Jenis-Jenis Linked List

1. Singly Linked List

Setiap node hanya memiliki satu pointer ke node berikutnya.

    10 → 20 → 30 → None

2. Doubly Linked List

Setiap node memiliki pointer ke node sebelumnya dan berikutnya.

    None ← 10 ⇄ 20 ⇄ 30 → None

3. Circular Linked List

Node terakhir menunjuk kembali ke node pertama.

    10 → 20 → 30
    ↑         ↓
    └─────────┘

4. Operasi Dasar Linked List

a. Traversal

Menampilkan seluruh isi linked list.

b. Insertion

Menambahkan node:

        Di awal
        Di tengah
        Di akhir

c. Deletion

Menghapus node tertentu.

d. Search

Mencari data dalam linked list.

## Ringkasan

| Konsep               | Penjelasan                             |
| -------------------- | -------------------------------------- |
| Node                 | Elemen yang menyimpan data dan pointer |
| Head                 | Node pertama                           |
| Singly Linked List   | Satu pointer ke node berikutnya        |
| Doubly Linked List   | Dua pointer (prev dan next)            |
| Circular Linked List | Node terakhir menunjuk ke node pertama |
| Insert               | Menambah data                          |
| Delete               | Menghapus data                         |
| Search               | Mencari data                           |
| Traversal            | Menampilkan seluruh data               |


Linked List adalah struktur data yang menyimpan kumpulan node yang saling terhubung menggunakan pointer. Struktur ini sangat cocok digunakan ketika ukuran data sering berubah karena proses penambahan dan penghapusan data dapat dilakukan dengan lebih efisien dibandingkan array.

```python
## Contoh Program Singly Linked List (Python)
# Membuat Node
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Membuat Linked List
class LinkedList:
    def __init__(self):
        self.head = None

    # Menambah data di akhir
    def append(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node

    # Menampilkan data
    def display(self):
        current = self.head

        while current:
            print(current.data, end=" -> ")
            current = current.next

        print("None")


# Program Utama
ll = LinkedList()

ll.append(10)
ll.append(20)
ll.append(30)

ll.display()
```

```python
## Contoh Program Insert di Awal
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def insert_awal(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def display(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")


ll = LinkedList()

ll.insert_awal(30)
ll.insert_awal(20)
ll.insert_awal(10)

ll.display()
```
