```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class OrderedList:
    def __init__(self):
        self.head = None

    def insert(self, item):
        newNode = Node(item)

        # jika list kosong
        if self.head is None:
            self.head = newNode
            return

        temp = self.head
        prev = None

        while temp:

            # jika ditemukan → sisipkan setelah
            if temp.data == item:
                newNode.next = temp.next
                temp.next = newNode
                return

            # jika item lebih kecil → berhenti
            if item < temp.data:
                break

            prev = temp
            temp = temp.next

        # sisipkan sebelum temp
        if prev is None:
            newNode.next = self.head
            self.head = newNode
        else:
            prev.next = newNode
            newNode.next = temp

    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end=" , ")
            temp = temp.next

        print("None")


# Program utama
olist = OrderedList()

olist.insert(3)
olist.insert(7)
olist.insert(10)
olist.insert(15)

print("Sebelum:")
olist.display()

olist.insert(8)

print("Sesudah:")
olist.display()
```
