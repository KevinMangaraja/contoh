# Infix, Prefix and Postfix Expressions
A. Infix (Notasi Manusia)
Format di mana operator diletakkan di antara dua operand. Ini adalah format yang kita gunakan sehari-hari.
- Format: [Operand] [Operator] [Operand]
- Contoh: A + B atau 5 * 3
- Masalah bagi Komputer: Komputer sulit membaca Infix karena harus memikirkan tanda kurung () dan urutan kekuatan operator (perkalian harus didahulukan dari penjumlahan).

B. Prefix (Notasi Polandia / Polish Notation)
Format di mana operator diletakkan sebelum atau di depan operand.
- Format: [Operator] [Operand] [Operand]
- Contoh: + A B atau * 5 3

C. Postfix (Reverse Polish Notation)
Format di mana operator diletakkan setelah atau di belakang operand. Notasi ini paling ideal bagi komputer karena tidak butuh tanda kurung dan dieksekusi linear dari kiri ke kanan.
- Format: [Operand] [Operand] [Operator]
- Contoh: A B + atau 5 3 *

        Tabel Perbandingan & Contoh Konversi

        Infix (Biasa)      |        A + B  |  A + B * C              | (A + B) *  C               | 

        Prefix (Depan)     |        + A B  |  + A * BC               | * + A B   C                |  

        Postfix (Belakang) |        A B +  |  A B C * +              | A B + C *                  |

        Keterangan         | operasi dasar |  B * C dihitung duluan  | Tanda kurung diprioritaskan|

contoh code : 

```python
def evaluasi_postfix(ekspresi):
    stack = []
    elemen = ekspresi.split()
    
    for karakter in elemen:
        if karakter.isdigit():
            stack.append(int(karakter))
        else:
            angka_kedua = stack.pop()
            angka_pertama = stack.pop()
            
            if karakter == '+':
                hasil = angka_pertama + angka_kedua
            elif karakter == '-':
                hasil = angka_pertama - angka_kedua
            elif karakter == '*':
                hasil = angka_pertama * angka_kedua
            elif karakter == '/':
                hasil = int(angka_pertama / angka_kedua)
                
            stack.append(hasil)
            
    return stack[0]

ekspresi_soal = "5 3 + 2 *"
print(evaluasi_postfix(ekspresi_soal))
```
