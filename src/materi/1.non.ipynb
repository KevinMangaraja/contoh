# Stack and queue

1. STACK (Tumpukan)
Stack adalah struktur data linear yang bekerja dengan prinsip LIFO (Last In, First Out). Data yang terakhir dimasukkan adalah yang pertama kali dikeluarkan.

- Analogi Visual: Tumpukan piring atau tumpukan buku.
- Karakteristik: Hanya memiliki satu pintu akses (ujung atas/Top). Semua proses menambah dan mengambil data dilakukan di ujung yang sama.
- Operasi Utama:
    a.Push: Menambahkan data baru ke posisi paling atas.
    b. Pop: Mengambil/menghapus data dari posisi paling atas.

- Contoh Fungsi: Fitur Undo (Ctrl+Z) dan tombol Back pada browser.

2. QUEUE (Antrean) & Operasinya
Queue adalah struktur data linear yang bekerja dengan prinsip FIFO (First In, First Out). Data yang pertama kali masuk akan menjadi yang pertama kali keluar.
Berbeda dengan Stack, Queue memiliki dua pintu akses:
- Rear/Back (Belakang): Tempat data baru masuk.
- Front/Head (Depan): Tempat data lama keluar.
Di dalam Queue, aktivitas menambah dan menghapus data memiliki istilah khusus, yaitu Enqueue dan Dequeue:

A. Enqueue (Proses Masuk)
- Definisi: Operasi untuk memasukkan atau menambahkan elemen data baru ke dalam antrean.
- Aturan Posisi: Data baru selalu dimasukkan dari ujung paling belakang (Rear/Back).
- Analogi: Seseorang yang baru datang dan langsung berdiri di barisan paling belakang antrean.

B. Dequeue (Proses Keluar)
- Definisi: Operasi untuk mengambil atau menghapus elemen data dari dalam antrean.
- Aturan Posisi: Data yang diambil selalu data yang berada di urutan paling depan (Front/Head).- Analogi: Orang di barisan paling depan yang selesai dilayani oleh kasir, lalu meninggalkan antrean.

```python
class Stack:
    def __init__(self):
        self.items = []
    def isEmpty(self):
        return self.items == []

    def push(self, item):

        self.items.append(item)

    def pop(self):

        return self.items.pop()

    def peek(self):

        return self.items[len(self.items)-1]

    def size(self):

        return len(self.items)

s=Stack()

s.push('hello')

s.push('true')


print(s.peek())

while not s.isEmpty():
    print(s.pop())
```
