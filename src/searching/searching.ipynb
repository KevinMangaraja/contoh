# Searching (pencarian) 
dalam struktur data adalah proses menemukan lokasi atau posisi elemen data tertentu di dalam sebuah kumpulan data (seperti array, list, atau tree). Proses ini mengembalikan indeks/posisi elemen jika ditemukan, atau memberikan indikasi jika data tidak ada.

Secara umum, algoritma pencarian dibagi menjadi dua kategori utama berdasarkan kondisi datanya: Linear Search (untuk data acak) dan Binary Search (untuk data yang sudah terurut).
1. Linear Search (Pencarian Berurutan)
Algoritma ini memeriksa setiap elemen satu per satu dari awal hingga akhir sampai data yang dicari ditemukan.
- Kondisi Data: Bebas (bisa acak atau terurut).
- Kompleksitas Waktu: \(O(n)\) — lambat untuk data skala besar.
- Cara Kerja: Mulai dari indeks 0, cocokkan data. Jika tidak cocok, pindah ke indeks berikutnya.

Contoh Kode Python (Linear Search)

```python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i 
    return -1  # Kembalikan -1 jika tidak ditemukan

data_acak = [10, 23, 4, 85, 32]
target_cari = 85

hasil = linear_search(data_acak, target_cari)
print(f"Linear Search: Element ditemukan pada indeks {hasil}")
```

2. Binary Search (Pencarian Biner)
Algoritma ini bekerja dengan membagi dua ruang pencarian secara berulang. Algoritma ini jauh lebih cepat daripada Linear Search.
- Kondisi Data: Wajib terurut (secara ascending atau descending).
- Kompleksitas Waktu: \(O(\log n)\) — sangat cepat untuk data skala besar.
- Cara Kerja: Cari titik tengah (mid). Jika target lebih kecil dari mid, persempit pencarian ke sebelah kiri. Jika target lebih besar, persempit ke sebelah kanan.Contoh Kode Python (Binary Search)

```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        
        if arr[mid] == target:
            return mid  
        elif arr[mid] < target:
            low = mid + 1  
        else:
            high = mid - 1  
            
    return -1 
data_terurut = [4, 10, 23, 32, 85]
target_cari = 85

hasil = binary_search(data_terurut, target_cari)
print(f"Binary Search: Element ditemukan pada indeks {hasil}")
```
